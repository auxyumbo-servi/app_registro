# 🎨 Tailwind CSS en NiceGUI: clases principales

En NiceGUI, el método:

```python
classes('...')
```

permite aplicar clases de **Tailwind CSS** para modificar estilos de los componentes.

---

# 🎨 Fondo (Background)

Define el color de fondo del elemento.

```python
classes('bg-yellow-300')
classes('bg-blue-500')
classes('bg-red-600')
```

## Formato

```
bg-{color}-{nivel}
```

## Ejemplos

```python
bg-white
bg-black
bg-gray-200
bg-green-500
bg-purple-700
```

Los niveles van normalmente de:

```
50  100  200  300  400  500  600  700  800  900  950
```

- `50` → muy claro
- `500` → color base
- `900` → muy oscuro

---

# ✏️ Color del texto

Define el color del texto.

```python
classes('text-black')
classes('text-white')
classes('text-gray-700')
```

## Formato

```
text-{color}-{nivel}
```

## Ejemplo

```python
ui.label('Hola').classes('text-yellow-500')
```

# 📏 Tamaño del texto

# 🔤 Peso de letra (Font Weight)

Controla el grosor de la fuente.

```text
font-thin
font-light
font-normal
font-medium
font-semibold
font-bold
font-extrabold
font-black
```

## Ejemplo

```python
ui.label('Importante').classes('font-bold')
```

---

# 📐 Espaciado

## Padding interno

Espacio dentro del componente.

```text
p-1
p-2
p-4
p-8
```

Ejemplo:

```python
ui.button('OK').classes('p-4')
```

---

## Padding por dirección

```text
px-4   # izquierda + derecha

py-2   # arriba + abajo

pt-4   # arriba

pb-4   # abajo

pl-4   # izquierda

pr-4   # derecha
```

---

## Margen externo

Espacio fuera del componente.

```text
m-4
mx-4
my-2
mt-8
mb-4
ml-4
mr-4
```

Ejemplo:

```python
ui.label('Hola').classes('mb-4')
```

---

# 🖱️ Estados (Hover, Focus)

Permiten cambiar estilos según acciones del usuario.

## Hover

Cuando el mouse pasa encima:

```python
classes(
    'bg-yellow-400 hover:bg-yellow-600'
)
```

Otros ejemplos:

```text
hover:text-white
hover:bg-blue-600
hover:scale-105
```

---

## Focus

Cuando el elemento está seleccionado:

```text
focus:ring
focus:outline-none
```

---

# 🚀 Ejemplo completo: botón moderno

```python
ui.button('Guardar').classes(
    '''
    bg-yellow-400
    text-black
    font-bold
    px-6
    py-3
    rounded-xl
    shadow-lg
    hover:bg-yellow-500
    '''
)
```

---

# 📌 Resumen rápido

| Clase | Función |
|---|---|
| `bg-*` | Fondo |
| `text-*` | Color y tamaño del texto |
| `font-*` | Grosor de letra |
| `p-*` | Espacio interno |
| `m-*` | Espacio externo |
| `rounded-*` | Bordes redondeados |
| `border-*` | Bordes |
| `shadow-*` | Sombras |
| `flex-*` | Distribución |
| `w-*` / `h-*` | Tamaño |
| `hover-*` | Efectos al pasar mouse |
| `focus-*` | Estados de selección |

---

## Ejemplo de tarjeta completa

```python
ui.card().classes(
    '''
    bg-white
    w-full
    p-6
    rounded-xl
    shadow-lg
    border
    border-gray-200
    '''
)
```

Con estas clases puedes construir la mayoría de interfaces modernas en NiceGUI.
